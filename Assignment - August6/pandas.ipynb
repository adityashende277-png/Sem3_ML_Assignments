```python
sales = np.array([
    120, 135, 98,
    160, 175, 190,
    88, 92, 105,
    210, 198, 230
])

quarters = sales.reshape(4, 3)

averages = quarters.mean(axis=1)

highest_quarter = np.argmax(averages)

print("Quarters:")
print(quarters)

print("Quarter averages:", averages)
print("Highest quarter index:", highest_quarter)
print("Highest average:", averages[highest_quarter])
```

```python
x = np.array([
    [50, 1200, 3],
    [80, 1500, 9],
    [65, 900, 5],
    [95, 2000, 7]
])

col_min = x.min(axis=0)
col_max = x.max(axis=0)

normalized = (x - col_min) / (col_max - col_min)

print(normalized)

print("Column minimums:", normalized.min(axis=0))
print("Column maximums:", normalized.max(axis=0))
```

```python
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(
    scores >= 90,
    "Gold",
    np.where(
        scores >= 75,
        "Silver",
        np.where(
            scores >= 60,
            "Bronze",
            "None"
        )
    )
)

print(medals)

print("Gold:", np.sum(medals == "Gold"))
print("Silver:", np.sum(medals == "Silver"))
print("Bronze:", np.sum(medals == "Bronze"))
print("None:", np.sum(medals == "None"))
```

```python
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)

print("Hours shape:", hours_col.shape)
print("Attendance shape:", attendance_col.shape)
table = np.hstack((hours_col, attendance_col))

print("Table shape:", table.shape)
totals = table.sum(axis=1, keepdims=True)

print("Totals shape:", totals.shape)
three_col = np.hstack((table, totals))

print("Three-column shape:", three_col.shape)
print(three_col)
w = np.array([0.5, 0.2, 0.3])

scores = three_col @ w

print("Scores shape:", scores.shape)
print("Scores:", scores)
average = scores.mean()

result = np.where(scores > average, 1, 0)

print("Average:", average)
print("Final result:", result)
print("Final shape:", result.shape)
```

```python
np.random.seed(7)

temps = np.random.randint(15, 41, 28)

weeks = temps.reshape(4, 7)

weekly_average = weeks.mean(axis=1)

hottest_week = np.argmax(weekly_average)

days_above_35 = np.sum(temps > 35)

print("Weeks:")
print(weeks)

print("\nWeekly averages:")
print(weekly_average)

print("\nHottest week index:", hottest_week)

print("Hottest week average:", weekly_average[hottest_week])

print("Days above 35:", days_above_35)
```

```python
np.random.seed(10)

A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

AB = A @ B
BA = B @ A

print("A:")
print(A)

print("\nB:")
print(B)

print("\nA @ B:")
print(AB)

print("\nB @ A:")
print(BA)

print("\nAre they equal?", np.array_equal(AB, BA))
```

```python
marks = np.array([
    [45, 28, 67, 80],
    [33, 90, 55, 41],
    [72, 66, 30, 58],
    [88, 79, 91, 84],
    [25, 40, 38, 52]
])
marks = np.maximum(marks, 35)
totals = marks.sum(axis=1)
topper_index = np.argmax(totals)

print("Updated marks:")
print(marks)

print("\nTotals:")
print(totals)

print("\nTopper index:", topper_index)
print("Topper total:", totals[topper_index])
```
