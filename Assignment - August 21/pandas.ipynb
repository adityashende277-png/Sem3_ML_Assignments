```python
import pandas as pd

# ==========================================
# LOAD DATA
# ==========================================

t = pd.read_csv("tips.csv")
titanic = pd.read_csv("Titanic-Dataset.csv")
```

# ==========================================
# PART A - tips.csv
# ==========================================

```python
# A1
t["tip_percent"] = (
    t["tip"] / t["total_bill"] * 100
).round(1)

print("A1:")
print(t[["total_bill", "tip", "tip_percent"]].head(5))
```

```python
# ------------------------------------------
# A2
# ------------------------------------------

print("\nA2(a): 10th row total_bill")
print(t.iloc[9, 0])

print("\nA2(b): Last 3 rows, first 2 columns")
print(t.iloc[-3:, :2])
```

```python
# ------------------------------------------
# A3
# ------------------------------------------

print("\nA3:")

result = t.loc[20:25, "sex":"size"]

print(result)

print("\nNumber of rows:", len(result))

# loc includes both 20 and 25,
# so rows 20,21,22,23,24,25 = 6 rows.
```

```python
# ------------------------------------------
# A4
# ------------------------------------------

print("\nA4(a): Bills not at Dinner")

print(
    (~(t["time"] == "Dinner")).sum()
)

print("\nA4(b): Party >= 3 AND tip > 4")

print(
    ((t["size"] >= 3) & (t["tip"] > 4)).sum()
)
```

```python
# ------------------------------------------
# A5
# ------------------------------------------

t["service"] = "normal"

t.loc[
    t["tip_percent"] > 20,
    "service"
] = "generous"

print("\nA5:")
print(t["service"].value_counts())
```

# ==========================================
# PART B - Titanic-Dataset.csv
# ==========================================

```python
# ------------------------------------------
# B1
# ------------------------------------------

print("\nB1: Passengers at each port")

port_counts = titanic["Embarked"].value_counts()

print(port_counts)

print("Total:", port_counts.sum())
```

```python
# ------------------------------------------
# B2
# ------------------------------------------

print("\nB2:")

women_third = titanic[
    (titanic["Sex"] == "female") &
    (titanic["Pclass"] == 3)
]

print("Women in third class:")
print(len(women_third))

print("Those who survived:")
print(
    (women_third["Survived"] == 1).sum()
)
```

```python
# ------------------------------------------
# B3
# ------------------------------------------

print("\nB3:")

high_fare = titanic.loc[
    titanic["Fare"] > 200,
    ["Name", "Age", "Fare"]
]

print(high_fare)

print("Number of passengers:")
print(len(high_fare))
```

```python
# ------------------------------------------
# B4
# ------------------------------------------

print("\nB4:")

first_class = titanic[
    titanic["Pclass"] == 1
]

third_class = titanic[
    titanic["Pclass"] == 3
]

first_percentage = (
    first_class["Survived"].mean() * 100
)

third_percentage = (
    third_class["Survived"].mean() * 100
)

print("First class survival percentage:")
print(first_percentage)

print("Third class survival percentage:")
print(third_percentage)
```

```python
# ------------------------------------------
# B5
# ------------------------------------------

titanic["age_group"] = "adult"

titanic.loc[
    titanic["Age"] < 18,
    "age_group"
] = "child"

print("\nB5:")
print(titanic["age_group"].value_counts())

# Missing Age values remain adult because
# everyone starts as adult and only Age < 18
# is changed to child.
```
